# Filtering Dataset NASA

1. Filter Request URI mengambil HTMLnya saja
2. Filter Status mengambil nilai yg 200
3. Filter Request Method mengambil yg GET
4. Merubah Request Time menjadi tanggal dan waktunya saja
5. Filter Berdasarkan IP Address(Remote host) dan User Agent(Request Protocol)

```python
import pandas as pd

# Upload file
from google.colab import files
uploaded = files.upload()

# Membaca file CSV hasil upload
df = pd.read_csv(
    next(iter(uploaded)),
    sep=None,
    engine='python',
    on_bad_lines='skip'
)


print("Data berhasil dimuat:", df.shape)
df.head()
```

```python
df.columns
```

## FILTER REQUEST URI YANG .html SAJA

```python
print("Jumlah data sebelum filter:", len(df))

# Filter URL yang berakhiran .html
df_html = df[df['url'].str.endswith('.html', na=False)]

print("Jumlah data sesudah filter:", len(df_html))
df_html.head()
```

Langkah kode tersebut memiliki fungsi sebagai berikut:

- len(df) menghitung jumlah baris keseluruhan sebelum proses filter sehingga kita tahu ukuran data mentah.

- df['url'].str.endswith('.html', na=False) merupakan proses inti filtering yang mengecek apakah sebuah URL benar-benar berakhiran .html. Jika iya, baris tersebut dipertahankan.

- na=False memastikan bahwa URL yang kosong atau nilai NaN tidak menyebabkan error dan otomatis dianggap tidak sesuai filter.

- Hasil filtering disimpan pada df_html, lalu kode menampilkan jumlah baris setelah filter dan beberapa contoh baris pertama menggunakan head().

Dari output yang dihasilkan:

- Jumlah data sebelum filter: 1,048,575 baris

- Jumlah data sesudah filter (hanya URL .html): 220,905 baris

- Hasil ini menunjukkan bahwa hanya sekitar 21% dari seluruh request yang benar-benar mengarah ke halaman HTML. Baris-baris yang ditampilkan pada df_html.head() memperlihatkan contoh URL seperti /ksc.html, /shuttle/missions/missions.html, dan lain-lain, yang berarti filter berjalan sesuai harapan.


## FILTER STATUS CODE 200

```python
print("Jumlah data sebelum filter:", len(df))

# Filter data dengan status response == 200
df_status200 = df[df['response'] == 200]

print("Jumlah data sesudah filter:", len(df_status200))
df_status200.head()
```

Dari output yang dihasilkan:

- Jumlah data sebelum filter: 1,048,575 baris

- Jumlah data sesudah filter (status 200): 951,345 baris

Artinya, sekitar 90,8% request berhasil diproses dengan baik oleh server. Ini menunjukkan bahwa sebagian besar permintaan yang diterima server tidak mengalami error seperti 404 (Not Found) atau 500 (Server Error).

## FILTER REQUEST METHOD GET

```python
print("Jumlah data sebelum filter:", len(df))

# Filter baris yang method-nya GET
df_get = df[df['method'] == 'GET']

print("Jumlah data sesudah filter:", len(df_get))
df_get.head()
```

Penjelasan bagian kode:

- len(df) menampilkan jumlah data sebelum difilter.

- df[df['method'] == "GET"] mengambil hanya baris yang kolom method-nya bernilai "GET". Hasilnya disimpan dalam variabel df_get.

- len(df_get) menunjukkan total baris setelah proses filter GET.

- .head() menampilkan beberapa baris pertama dari data GET.

Hasil Output

- Jumlah data sebelum filter: 1,048,575 baris

- Jumlah data sesudah filter: 1,043,273 baris

Berarti sebagian besar request di dataset ini menggunakan metode GET, yaitu sekitar 99.49% dari keseluruhan data. Ini normal karena GET adalah metode default untuk mengambil resource.

## MENGUBAH REQUEST TIME MENJADI TANGGAL & WAKTU

```python
import pandas as pd

print("Jumlah data sebelum transformasi waktu:", len(df))

# Konversi kolom time menjadi datetime
df['time'] = pd.to_datetime(df['time'], errors='coerce')

# Pisahkan menjadi tanggal dan waktu saja
df['tanggal'] = df['time'].dt.date
df['waktu'] = df['time'].dt.time

print("Jumlah data sesudah transformasi (baris sama):", len(df))

df[['time', 'tanggal', 'waktu']].head(10)
```

## FILTER BERDASARKAN IP ADDRESS & REQUEST PROTOCOL

```python
print("Jumlah data sebelum filter:", len(df))

# ---------------------- FILTER 1: BERDASARKAN IP ADDRESS ----------------------
ip_target = "103.10.20.30"   # ganti IP sesuai yang kamu mau

df_ip = df[df['host'] == ip_target]

print("\n--- FILTER BERDASARKAN IP ADDRESS ---")
print("IP yang difilter :", ip_target)
print("Jumlah data sesudah filter IP :", len(df_ip))
print(df_ip.head())

# ---------------------- FILTER 2: BERDASARKAN REQUEST PROTOCOL ----------------------
# contoh: hanya ambil URL yang menggunakan 'https://'
df_protocol = df[df['url'].str.startswith("https://", na=False)]

print("\n--- FILTER BERDASARKAN REQUEST PROTOCOL (HTTPS) ---")
print("Jumlah data sesudah filter protokol :", len(df_protocol))
print(df_protocol.head())
```

Pada tahap ini dilakukan dua proses penyaringan data:
1. penyaringan berdasarkan alamat IP tertentu, dan
2. penyaringan berdasarkan protokol request (misalnya HTTPS).

Data awal memiliki 1.048.575 baris, namun setelah difilter berdasarkan IP 103.10.20.30, jumlah baris menjadi 0, karena IP tersebut tidak terdapat pada dataset. Selanjutnya, penyaringan berdasarkan request protocol seperti “https://” juga menghasilkan 0 data, karena dataset NASA 1995 ini hanya menggunakan protokol HTTP, sehingga tidak ada baris yang memenuhi syarat.

Dengan demikian, kedua filter menghasilkan data kosong, namun proses filtering telah berhasil dilakukan sesuai instruksi.
