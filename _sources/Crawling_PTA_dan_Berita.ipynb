## Link Keluar

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def crawl_links_teknik(pages):
    data = []
    no = 1

    for page in pages:
        try:
            response = requests.get(page)
            response.raise_for_status()
            soup = BeautifulSoup(response.content, 'html.parser')

            # Ambil semua link
            links = soup.find_all('a', href=True)
            for link in links:
                href = link['href']
                teks = link.get_text(strip=True)

                # Filter hanya link keluar (mulai dengan http & bukan link internal teknik.trunojoyo.ac.id)
                if href.startswith("http") and "teknik.trunojoyo.ac.id" not in href:
                    data.append({
                        "No": no,
                        "Page": page,
                        "Link Keluar": href,
                        "Teks": teks if teks else "-"
                    })
                    no += 1

        except requests.exceptions.RequestException as e:
            print(f"Gagal mengakses {page}: {e}")

    # Buat DataFrame
    df = pd.DataFrame(data)

    print("\n🔗 Daftar Link Keluar Fakultas Teknik:")
    print(df.to_string(index=False))

    # Simpan hasil ke CSV
    df.to_csv("link_keluar_fakultas_teknik.csv", index=False, encoding="utf-8-sig")
    print("\n✅ Data link keluar berhasil disimpan ke link_keluar_fakultas_teknik.csv")

    return df


# Daftar halaman Fakultas Teknik yang ingin diambil link keluarnya
pages_to_crawl = [
    "https://teknik.trunojoyo.ac.id/",
    "https://teknik.trunojoyo.ac.id/sekilas/",
    "https://teknik.trunojoyo.ac.id/berita/",
    "https://teknik.trunojoyo.ac.id/prodi/",
]

# Jalankan
crawl_links_teknik(pages_to_crawl)
```

## stemming dan lematisasi

```python
# 1. Install library
!pip install Sastrawi

# 2. Import
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# 3. Buat stemmer
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# 4. Baca dataset
file_path = "/content/drive/MyDrive/PPW/pta_fatek_3halaman.csv"
df = pd.read_csv(file_path, encoding='utf-8')

# 5. Ambil kolom abstrak
df = df[['judul', 'prodi', 'abstrak']]

# 6. Fungsi stemming
def apply_stemming(text):
    return stemmer.stem(str(text))

# 7. Terapkan stemming
df['abstrak_stemming'] = df['abstrak'].apply(apply_stemming)

# 8. Lematisasi sederhana (manual, opsional)
# Bahasa Indonesia tidak punya lemmatizer default seperti bahasa Inggris,
# tapi kita bisa bikin kamus dasar untuk beberapa kata umum.
lemma_dict = {
    "berlari": "lari",
    "makanan": "makan",
    "minuman": "minum",
    "bermain": "main",
    "tertidur": "tidur",
    "menjalankan": "jalan"
}

def apply_lemmatization(text):
    words = text.split()
    lemmas = [lemma_dict.get(word, word) for word in words]
    return " ".join(lemmas)

df['abstrak_lematisasi'] = df['abstrak_stemming'].apply(apply_lemmatization)

# 9. Simpan hasil
output_path = "/content/drive/MyDrive/PPW/pta_fatek_stemming_lemma.csv"
df.to_csv(output_path, index=False, encoding='utf-8-sig')

# 10. Tampilkan contoh hasil
print("✅ Proses stemming dan lemmatisasi selesai.")
print(df[['abstrak', 'abstrak_stemming', 'abstrak_lematisasi']].head())
```

## stemming dan lematisasi

```python
# 1. Import
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import re

# 2. Buat stemmer
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# 3. Baca dataset
file_path = "/content/drive/MyDrive/PPW/CrawlBerita_Kesehatan-Pariwisata.csv"

df_berita = pd.read_csv(file_path, encoding='utf-8')
print(f"✅ File berita ditemukan! Total entri: {len(df_berita)}")

# 4. Bersihkan & stemming
def bersihkan_text(teks):
    if pd.isna(teks):
        return ""
    teks = str(teks)
    teks = re.sub(r'[^a-zA-Z0-9\s]', ' ', teks)  # hapus tanda baca
    teks = re.sub(r'\s+', ' ', teks)  # hapus spasi berlebih
    return teks.strip().lower()

def apply_stemming(teks):
    teks_bersih = bersihkan_text(teks)
    return stemmer.stem(teks_bersih)

if 'isi_berita' in df_berita.columns:
    df_berita['isi_berita_bersih'] = df_berita['isi_berita'].apply(bersihkan_text)
    df_berita['isi_berita_stemming'] = df_berita['isi_berita_bersih'].apply(apply_stemming)

    # 5. Placeholder lemmatisasi (sama dengan hasil stemming)
    df_berita['isi_berita_lematisasi'] = df_berita['isi_berita_stemming']

    # 6. Simpan hasil
    output_path = "/content/drive/MyDrive/PPW/berita_stemming_lemma.csv"
    df_berita.to_csv(output_path, index=False, encoding='utf-8-sig')

    # 7. Tampilkan 10 baris pertama hasilnya
    print("\n📌 Contoh hasil stemming & lemmatisasi:")
    print(df_berita[['isi_berita', 'isi_berita_stemming', 'isi_berita_lematisasi']].head(10))
else:
    print("⚠️ Kolom 'isi_berita' tidak ditemukan dalam file CSV.")
```
